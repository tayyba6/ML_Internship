```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Pages:", len(df))
print("Columns:", len(df.columns))

declining_rate = (df["trend_direction"] == "down").mean()
print("Declining pages:", round(declining_rate, 3))
```

The starter dataset contains 30,000 pages and 44 columns. About 54.2% of the pages have a downward trend direction, which provides a useful signal to investigate when building a ranking for review prioritization. These numbers make the content-review prioritization problem worth exploring further.
